## 1. Contexto y motivación

En Ecuador, el estado de los embalses hidroeléctricos (cotas, en metros sobre el nivel del mar) es un tema de interés público recurrente: racionamientos, apagones y crisis de generación se comunican por vías oficiales. La ciudadanía carece normalmente de una forma **sencilla, reproducible e independiente** de verificar qué dicen los registros medidos.

Este proyecto NO afirma que exista desinformación. Su premisa es neutral:

> Poner junto, en una misma línea de tiempo verificable, **lo que dicen los sensores oficiales** y **lo que dicen los comunicados oficiales**, y dejar que cualquier persona saque sus propias conclusiones.

## 2. Objetivos del proyecto

| # | Objetivo | Criterio de éxito |
|---|----------|-------------------|
| O1 | Recolectar diariamente la cota de los embalses Mazar, Amaluza y Sopladora desde la fuente oficial (CELEC SUR) | Serie continua sin intervención manual, con evidencia archivada |
| O2 | Presentar cada serie contra sus umbrales operativos de referencia (gráfico semáforo) | Cualquier lector entiende en < 10 segundos la posición del nivel |
| O3 | Registrar comunicados oficiales sobre embalses y ubicarlos sobre la serie medida | Cada comunicado tiene una cota medida asociada automática |
| O4 | Documentar limitaciones, fuentes y método con total transparencia | Un tercero puede reproducir todo el pipeline desde cero |

## 3. Alcance

- **Embalses monitorizados:** Mazar, Amaluza (central Molino) y Sopladora — cascada del río Paute, operada por CELEC SUR.
- **Variable principal:** cota diaria del embalse (msnm), medida a medianoche hora Ecuador.
- **Variables de contexto:** fechas de comunicados oficiales registrados manualmente en `data/comunicados/comunicados.csv`.
- **Fuente primaria:** la misma API pública que alimenta el tablero oficial [Gráficas de Producción de CELEC SUR](https://generacioncsr.celec.gob.ec/graficasproduccion/).
- **Fuera de alcance:** interpretabilidad de decisiones operativas, proyecciones hidrológicas, asignación de responsabilidades.

## 5. Criterios de éxito globales

1. **Neutralidad:** ningún gráfico ni tabla clasifica declaraciones como verdaderas o falsas; solo superpone datos.
2. **Trazabilidad:** cada dato lleva su fecha de consulta (`fecha_consulta`) y el identificador del punto de medición (`mrid`).
3. **Reproducibilidad:** el pipeline completo (recolección → preparación → figuras) se ejecuta con dos comandos documentados en el README.
4. **Persistencia:** intento diario de respaldo inmutable en web.archive.org (mejor esfuerzo, registrado en `data/raw/archivo_web.log`).

## 6. Riesgos y consideraciones éticas

| Riesgo | Mitigación |
|--------|------------|
| Malentendido del lector (interpretar umbrales como "bueno/malo") | La banda verde se rotula "banda normal", no "banda segura" |
| Sobre-extracción del servidor público | 1 llamada por embalse al día (3 en total), pausas de cortesía |
| Deriva de la fuente (cambio de API) | Evidencia archivada y README que documenta el descubrimiento del endpoint |
| Parcialidad en el registro de comunicados | Reglas explícitas en `data/comunicados/LEEME.md`: registrar declaraciones de cualquier signo |
| Uso indebido del proyecto para afirmar fraude sin base | README y notebooks repiten el principio: los datos se presentan, no se concluye por el lector |
